```python
# complex number
z = 1 + 6j
# visit real part and imaginary part of a complex number
real_part = z.real
imaginary_part = z.imag

# conjugate and modulus
conjugate_result = z.conjugate()
modulus = abs(z)

# complex number calculation
w = 1 + 4j
sum_result = z + w  # addition
product_result = z * w  # multiplication

# print results
print("z:", z)
print("Real part of z:", real_part)
print("Imaginary part of z:", imaginary_part)
print("Conjugate part of z:", conjugate_result)
print("Modulus:", modulus)
print("w:", w)
print("Sum of z and w:", sum_result)
print("Product of z and w:", product_result)
```

```python
# open
file_obj = open('lines.txt', 'w+', encoding='utf-8')
lines = ['Line 1\n', 'Line 2\n', 'Line 3\n']
file_obj.writelines(lines)
file_obj.close()

with open('lines.txt', 'r', encoding='utf-8') as f:
    content = f.read()
    print(content , '*************')
with open('lines.txt', 'r', encoding='utf-8') as f:
    content = f.readlines()
    print(content , '\n*************')
with open('lines.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line, end='')  # end='' to avoid adding extra new line
```

```python
# offset and whence the file pointer
offset = 1
whence = 0  # 0: from the beginning of the file, 1: from the current position, 2: from the end of the file
with open('lines.txt', 'r', encoding='utf-8') as f:
    f.seek(offset, whence)
    content = f.read(8)
    print(content)
```

```python
# shallow copy: copy()
fruits = ['apple', 'banana', 'cherry', ['orange', 'grape']]
fruits_copy = fruits.copy()

fruits[0] = 'mango'
fruits[-1][0] = 'papay'
print(fruits)
print(fruits_copy)

# shallow copy creates a new list object, that shares the same memory reference for the nested list. So when we modify the nested list in the original list, it also modifies the nested list in the copied list.

# shallow copy: a[list] = b[list] + c[list]
list_a = [1, 2, 3, [4, 5, 6]]
double_list_a = list_a * 2
print(double_list_a)
list_a[-1][0] = 7
print(list_a)
print(double_list_a)

# deep copy
from copy import deepcopy

```

```python
# closures
# A closure is a function that can access and manipulate its external lexical environment. It allows the inner function to remember and access variables from the outer function even after the outer function has finished executing. Closures are often used to create functions with persistent state or to implement decorators. note: closure may cause memory leak if not used properly, because the inner function holds a reference to the outer function's variables, which may prevent them from being garbage collected.
def outer_function(x):
    def inner_function(y):
        return x + y

    return inner_function

closure = outer_function(10)
print(closure(5))

def outer():
    b = 10
    def inner(x):
        return b + 1 + x
    return inner

b = 2
f = outer()
print(f(b))
```
